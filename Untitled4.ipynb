**1. Introduction**

Due to their compact design, high throughput and well-defined residence time behaviour, tubular reactors are widely used in the chemical industry for the continuous production of bulk chemicals. Thermal operation mode is an important reactor design criterion, since the most technically relevant reactions are either exothermic or endothermic. The reaction rate couples mass and energy balances in a non-linear way, which allows analytical solutions only in highly idealized limiting cases, and in general, numerical methods are required.

This assignment deals with the exothermic hydrolysis of ethylene oxide (EO) with water to ethylene glycol (EG):

EO + H₂O ⟶ EG                                       (1)

Ethylene glycol is an important industrial chemical with an estimated worldwide production of 56.97 megatonnes in 2022 for use in polyester fibres, antifreeze formulations and PET production. The reaction is highly exothermic means -92.2KJ/mol , which means that the heat released during the reaction can cause a significant temperature rise inside the reactor, thus accelerating the reaction rate via the Arrhenius dependence.

The reaction is performed in a Plug Flow Tubular Reactor (PFTR), an idealized steady-state reactor model that assumes perfect radial mixing and no axial back-mixing. Mathematically, the PFTR is equivalent to a batch stirred tank reactor (bSTR) under this assumption. The reaction time tR in a batch reactor is exactly equal to the mean residence time τˉ in the PFTR.

The main objective of this work is to adapt the bSTR mathematical framework to the PFTR case, implement numerically the coupled material and energy balances, verify the implementation with rigorous analytical checks, and study the effect of key process and design parameters on EO conversion and temperature profile.

### 2.1 Starting Point: The Batch Reactor (bSTR) Equations

The assignment notebook lists the general bSTR equations as the mathematical starting point. For a single reaction A 1 → A 2 with constant density, the material balance in the bSTR is: The general material balance for the bSTR at constant density is:

$$\frac{d c_i}{d t} = \sum_{j=1}^{M} \nu_{i,j} \, r_j \tag{1}$$

where ci is the concentration of component i, vi,j is the stoichiometric coefficient of component i in the reaction j and rj is the rate of reaction j.

The corresponding energy balance for the batch reactor is:

$$V_R \, \rho \, c_p \frac{d T}{d t} = - V_R \sum_{j=1}^{M} \Delta_R H_j \, r_j - h_{\mathrm{wall}} A_{\mathrm{wall}} \left(T - \overline{T}_{\mathrm{wall}}\right) \tag{2}$$

This equation states that the reactor temperature changes because of two effects: heat released or absorbed by chemical reaction, and heat exchanged between the reactor contents and the wall.


The reaction rate follows an Arrhenius-based power-law:

$$r_j = k_{j,0} \exp \left(-\frac{E_{A,j}}{R \, T} \right) \prod_{i=1}^{N} c_i^{n_{i,j}} \tag{3}$$

Equations (1) to (3) form a coupled system of ordinary differential equations. In the batch reactor, these equations are solved as an initial value problem in time, using the initial concentrations and initial temperature at t = 0.

---

### 2.2 Key Adaptation: Space–Time Transformation

The key mathematical step in this assignment is the transformation of the batch-reactor equations into the equations of a plug flow tubular reactor. In a PFTR operated at steady state, the fluid moves through the tube with constant mean velocity
u, and the position in the reactor is described by the axial coordinate
z. The residence time of a fluid element at position $z$ in the PFTR is:

$$t = \frac{z}{u}, \qquad \text{therefore} \qquad dt = \frac{dz}{u} \tag{4}$$

This relation is called the space time transformation. It allows the time derivative in the batch reactor equations to be replaced by a derivative with respect to reactor length Substituting into the material balance gives the PFTR form:

$$\frac{d c_i}{d z} = \frac{1}{u} \sum_{j=1}^{M} \nu_{i,j} \, r_j \tag{5}$$

The same transformation is applied to the energy balance. As a result, the PFTR can also be treated as an initial value problem, but now in the axial coordinate
z instead of time
t. This is why the same numerical solution strategy used for the bSTR can also be used for the PFTR.

---

### 2.3 Reaction Kinetics for EO Hydrolysis

The reaction considered is:

$$\mathrm{EO} + \mathrm{H_2O} \rightarrow \mathrm{EG} \tag{6}$$

Because water is present in large excess, its concentration changes only slightly during the reaction and can be treated as effectively constant. Therefore, the rate law can be simplified to a pseudo-first-order expression with respect to ethylene oxide.

$$r = k(T) \, c_{\mathrm{EO}} \tag{7}$$

The temperature-dependent rate constant as given in the assignment:

$$k(T) = 5.01 \times 10^{6} \exp\left(-\frac{72500 \; \mathrm{J \, mol^{-1}}}{R \, T}\right) \; \mathrm{s^{-1}} \tag{8}$$

This means that the reaction rate depends on both concentration and temperature. Since the reaction is exothermic, an increase in temperature increases the rate constant, which accelerates the reaction further. This nonlinear coupling between mass and energy balances is the main reason why a numerical solution is required.

---

### 2.4 Material Balances in the PFTR

For the EO hydrolysis reaction, the stoichiometry is 1:1. One mole of EO is consumed and one mole of EG is formed. Applying the space time transformation to the material balances gives the PFTR equations.

$$u \, \frac{d c_{\mathrm{EO}}}{d z} = -r \tag{9}$$

$$u \, \frac{d c_{\mathrm{EG}}}{d z} = +r \tag{10}$$


Equation (9) describes the decrease in EO concentration along the reactor axis, while Equation (10) describes the increase in EG concentration. Since the reaction stoichiometry is one-to-one and density is assumed constant, the sum of both concentrations must remain constant throughout the reactor.



$$c_{\mathrm{EO}}(z) + c_{\mathrm{EG}}(z) = c_{\mathrm{EO,in}} \qquad \ \tag{11}$$

---

### 2.5 Energy Balance Adaptation for the Cylindrical PFTR

Applying the same space time transformation to the bSTR energy balance (Eq. 3), and adapting the wall-heat-transfer geometry to a cylindrical tube (surface to volume ratio for a cylinder), the PFTR energy balance becomes:

$$\frac{A_{\mathrm{wall}}}{V_R} = \frac{\pi \, d_R \, \Delta z}{\dfrac{\pi}{4} d_R^2 \, \Delta z} = \frac{4}{d_R} \tag{12}$$

The PFTR energy balance therefore becomes:

$$\rho \, c_p \, u \, \frac{dT}{dz} = \left(-\Delta_R H\right) r \; - \; h_{\mathrm{wall}} \frac{4}{d_R} \left(T - T_{\mathrm{wall}}\right) \tag{13}$$

The first term on the right-hand side represents the heat released by the exothermic reaction. The second term represents heat transfer through the tube wall to or from the surroundings. Since the reaction enthalpy is negative, The factor 4/dr  is a direct consequence of the cylindrical reactor geometry and is one of the most important modifications when adapting the bSTR equations to the PFTR case. In the batch-reactor example, the wall area and reactor volume were calculated directly from vessel geometry. In the PFTR, this ratio is built into the model through the surface to volume relation of the tube.
